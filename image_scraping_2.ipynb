```python
import os
import selenium
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import base64
import time
import urllib.request
import requests


#google images search bar
userSearch = input("Search bar:")
#userSearch = "hardhat"
#split search query then add "+" between every word
userSearch=userSearch.split()
userSearch = "+".join(userSearch)

GOOGLE_IMAGES = "https://www.google.com/search?q="+ userSearch +"&rlz=1C5CHFA_enGB937GB937&sxsrf=ALeKk02twWeFWJCESdRh27hZxC0iZwBS6w:1617724623722&source=lnms&tbm=isch&sa=X&ved=2ahUKEwiu0ans_envAhUkT98KHZ7CDiMQ_AUoAXoECAEQAw&biw=714&bih=732"
print(GOOGLE_IMAGES)

# Configure the web driver
options = webdriver.ChromeOptions()
#options.add_argument('--disable-extensions')
#options.add_argument('--headless')
#options.add_argument('--disable-gpu')
#options.add_argument('--no-sandbox')
#options.add_argument('--ignore-ssl-errors')
#options.add_argument('--ignore-certificate-errors')
#options.add_argument('--disable-blink-features=AutomationControlled')
driver = webdriver.Chrome(options=options)
driver.get(GOOGLE_IMAGES)
```

```python
scrollNumber=5
for i in range(scrollNumber):    
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(5)
    print('scroll done')
#driver.quit()    
```

```python
#get all image results
img_box = driver.find_elements(By.CLASS_NAME,'wXeWr')
print("image boxes: "+str(len(img_box)))
img_links=[]

#for evrey image element 
for img_ele in img_box[:-1]:
    img_ele.click()
    loader_Xpath = '//*[@id="Sva75c"]/div[2]/div[2]/div[2]/div[2]/c-wiz/div/div/div/div[3]/div[1]/div[1]'
    try:
        loader = driver.find_element(By.XPATH, loader_Xpath)
    #wait for image to load
        while not loader.value_of_css_property('display') == 'none':
            pass
    
    
        #get image URL
        img_inbox_Xpath = '//*[@id="Sva75c"]/div[2]/div[2]/div[2]/div[2]/c-wiz/div/div/div/div[3]/div[1]/a/img[1]'
        img_inbox = driver.find_element(By.XPATH, img_inbox_Xpath) 
        img_link = img_inbox.get_attribute('src')
        
        #if image cant load the source URL
        if img_link.startswith('data:image') or img_link.startswith('https://encrypted'):
            print("Thumbnail")
        
        #if the source URL found add it to img_links list
        else:
            print(img_link)
            img_links.append(img_link)
            
    except:
        print("different Xpath...")
```

```python
print("image links: "+str(len(img_links)))    

Download_counter = 0
img_counter = 1
SAVE_FOLDER = 'C:/Users/ASUS/GP/test_img/'
#SAVE_FOLDER = 'E:/Piechart/'

for image in img_links:
    try:
        response = requests.get(image, verify=False)
        if response.status_code == 200:
            open(SAVE_FOLDER + userSearch+ str(img_counter)+'.jpeg', 'wb').write(response.content)
            print("Image downloaded successfully!")
            Download_counter += 1
            
        else:
            print("Failed to download image.")
              
    except Exception as e:
        print("error in image number"+str(img_counter)+" :"+repr(e))
    
    img_counter += 1
    
print("Total downloaded images: "+str(Download_counter))        
```
